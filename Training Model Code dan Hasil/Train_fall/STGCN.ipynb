```python
# ============================================================================
#  SAPA — EKSPERIMEN PEMBANDING: "KEPALA JATUH" versi ST-GCN
#  Kerangka CV/metrik/callbacks IDENTIK train_fall.py (BiLSTM) -> perbandingan ADIL.
# ----------------------------------------------------------------------------
#  Tujuan: bukti "kematangan arsitektur" -> BiLSTM vs ST-GCN di data & split yang SAMA.
#  Input : ntu_fall.npz. Output (ke MODEL_DIR): fall_head_stgcn.pt + fall_stgcn_*.csv/json/png
#  Bandingkan fall_stgcn_cv_summary.json (ini) dgn fall_cv_summary.json (BiLSTM).
#
#  ST-GCN = skeleton diperlakukan sbg GRAF (sendi=titik, tulang=sisi) + konvolusi
#  spatio-temporal. Beda dgn BiLSTM yg mengabaikan struktur tubuh.
#  Copy tiap "# ===== CELL N =====" ke satu cell Colab. GPU disarankan. SEED sama = fold sama.
# ============================================================================


# ===== CELL 1: Import + Konfigurasi =====
import os, re, json, csv, time, numpy as np
import torch, torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score,
                             recall_score, matthews_corrcoef, cohen_kappa_score,
                             confusion_matrix, classification_report)
from sklearn.dummy import DummyClassifier
from tqdm.auto import tqdm
import pandas as pd
import matplotlib.pyplot as plt

BASE      = "/content/drive/MyDrive/Belajar AI/Compfest/Dataset"
NPZ_PATH  = f"{BASE}/preprocessed_2/ntu_fall.npz"
MODEL_DIR = f"{BASE}/models"; os.makedirs(MODEL_DIR, exist_ok=True)
TAG       = "stgcn"                        # penanda output (biar tak menimpa BiLSTM)

FALL_JOINTS  = list(range(5, 17)); USE_CHANNELS = 2
CLASS_NAMES  = ["normal", "oleng", "jatuh"]; N_CLASSES = 3
METRIC_COLS  = ["acc", "bal_acc", "macroF1", "weightedF1", "recall_jatuh", "mcc", "kappa"]

SEED, HIDDEN = 42, 64                       # ST-GCN: HIDDEN=channel dasar
LR, WEIGHT_DECAY, BATCH = 1e-3, 1e-4, 64
EPOCHS, PATIENCE, GRAD_CLIP = 60, 10, 5.0
LR_FACTOR, LR_PATIENCE = 0.5, 4
N_FOLDS, MONITOR = 5, "macroF1"

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
torch.manual_seed(SEED); np.random.seed(SEED)
print("Device:", DEVICE)


# ===== CELL 2: Load data + slice (4D, TANPA flatten) + grup performer =====
d = np.load(NPZ_PATH, allow_pickle=True)
kp = d["keypoints"]; y = d["labels"].astype(np.int64); meta = d["meta"].astype(str)
N = kp.shape[0]
print(f"Total jendela: {N} | distribusi: "
      f"{ {CLASS_NAMES[c]: int((y==c).sum()) for c in range(N_CLASSES)} }")

# ST-GCN butuh bentuk [N,T,V,C] -> JANGAN diflatten (beda dari BiLSTM)
X = kp[:, :, FALL_JOINTS, :USE_CHANNELS].astype(np.float32)     # [N,45,12,2]
print("X:", X.shape)
def performer(s):
    m = re.search(r'P(\d{3})', s); return m.group(0) if m else 'P000'
groups = np.array([performer(m) for m in meta])
print("jumlah performer:", len(set(groups)))


# ===== CELL 3: Model ST-GCN (graf sendi + konvolusi spatio-temporal) =====
def build_A(V, edges):
    A = np.eye(V, dtype=np.float32)
    for a, b in edges: A[a, b] = 1; A[b, a] = 1
    Dm = np.diag(A.sum(1) ** -0.5)
    return torch.tensor(Dm @ A @ Dm, dtype=torch.float32)       # A ternormalisasi (simetris)

# graf 12 sendi jatuh (COCO 5..16 diremap ke 0..11)
FALL_EDGES = [(0,2),(2,4),(1,3),(3,5),(0,1),(0,6),(1,7),(6,7),(6,8),(8,10),(7,9),(9,11)]
A_FALL = build_A(12, FALL_EDGES).to(DEVICE)

class GraphConv(nn.Module):
    def __init__(self, ci, co, A):
        super().__init__(); self.register_buffer('A', A); self.conv = nn.Conv2d(ci, co, 1)
    def forward(self, x):                      # x [N,C,T,V]
        return torch.einsum('nctv,vw->nctw', self.conv(x), self.A)

class STBlock(nn.Module):
    def __init__(self, ci, co, A, kt=9, stride=1, res=True):
        super().__init__()
        self.gcn = GraphConv(ci, co, A)
        self.tcn = nn.Sequential(nn.BatchNorm2d(co), nn.ReLU(),
                                 nn.Conv2d(co, co, (kt, 1), (stride, 1), ((kt-1)//2, 0)),
                                 nn.BatchNorm2d(co), nn.Dropout(0.2))
        if not res: self.res = lambda x: 0
        elif ci == co and stride == 1: self.res = nn.Identity()
        else: self.res = nn.Sequential(nn.Conv2d(ci, co, 1, (stride, 1)), nn.BatchNorm2d(co))
        self.relu = nn.ReLU()
    def forward(self, x): return self.relu(self.tcn(self.gcn(x)) + self.res(x))

class STGCN(nn.Module):
    def __init__(self, ci, n_classes, A, h=64):
        super().__init__(); V = A.shape[0]
        self.data_bn = nn.BatchNorm1d(ci * V)
        self.blocks = nn.ModuleList([STBlock(ci, h, A, res=False), STBlock(h, h, A),
                                     STBlock(h, h*2, A, stride=2), STBlock(h*2, h*2, A)])
        self.fc = nn.Linear(h*2, n_classes)
    def forward(self, x):                      # x [N,T,V,C]
        Ns, T, V, C = x.shape
        x = x.permute(0, 3, 2, 1).contiguous().view(Ns, C*V, T)
        x = self.data_bn(x).view(Ns, C, V, T).permute(0, 1, 3, 2).contiguous()  # [N,C,T,V]
        for b in self.blocks: x = b(x)
        return self.fc(x.mean(dim=(2, 3)))     # global pool T,V

def make_model(): return STGCN(USE_CHANNELS, N_CLASSES, A_FALL, HIDDEN).to(DEVICE)


# ===== CELL 4: Fungsi bantu + train 1 fold =====
def make_loader(idx, shuffle, drop_last=False):
    return DataLoader(TensorDataset(torch.tensor(X[idx]), torch.tensor(y[idx])),
                      batch_size=BATCH, shuffle=shuffle, drop_last=drop_last)

def full_metrics(yt, yp):
    return {"acc": accuracy_score(yt, yp), "bal_acc": balanced_accuracy_score(yt, yp),
            "macroF1": f1_score(yt, yp, average='macro', zero_division=0),
            "weightedF1": f1_score(yt, yp, average='weighted', zero_division=0),
            "recall_jatuh": recall_score(yt, yp, labels=[2], average='macro', zero_division=0),
            "mcc": matthews_corrcoef(yt, yp), "kappa": cohen_kappa_score(yt, yp)}

def predict(model, dl):
    model.eval(); ys, ps = [], []
    with torch.no_grad():
        for xb, yb in dl:
            ps.append(model(xb.to(DEVICE)).argmax(1).cpu().numpy()); ys.append(yb.numpy())
    return np.concatenate(ys), np.concatenate(ps)

def train_one_fold(fold, idx_tr, idx_va):
    cnt = np.bincount(y[idx_tr], minlength=N_CLASSES).astype(np.float32)
    cw = cnt.sum() / (N_CLASSES * np.clip(cnt, 1, None))
    model = make_model()
    criterion = nn.CrossEntropyLoss(weight=torch.tensor(cw, device=DEVICE))
    optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max',
                                                           factor=LR_FACTOR, patience=LR_PATIENCE)
    dl_tr = make_loader(idx_tr, True, drop_last=True)   # drop_last: hindari batch-1 (BatchNorm)
    dl_va = make_loader(idx_va, False)
    best, best_ep, wait, best_state, hist = -1.0, 0, 0, None, []
    for ep in range(1, EPOCHS + 1):
        model.train(); run = 0.0; seen = 0
        pbar = tqdm(dl_tr, desc=f"Fold {fold} Ep {ep:02d}/{EPOCHS}", leave=False)
        for xb, yb in pbar:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)
            optimizer.zero_grad(); loss = criterion(model(xb), yb); loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP)
            optimizer.step(); run += loss.item() * len(yb); seen += len(yb)
            pbar.set_postfix(loss=f"{loss.item():.4f}")
        yv, pv = predict(model, dl_va); mv = full_metrics(yv, pv)
        scheduler.step(mv[MONITOR]); lr_now = optimizer.param_groups[0]['lr']
        hist.append({"fold": fold, "epoch": ep, "train_loss": round(run/max(seen,1), 4),
                     **{k: round(v, 4) for k, v in mv.items()}, "lr": lr_now})
        print(f"  fold {fold} ep {ep:02d} | loss {run/max(seen,1):.4f} | "
              f"macroF1 {mv['macroF1']:.3f} | recall_jatuh {mv['recall_jatuh']:.3f} | lr {lr_now:.1e}")
        if mv[MONITOR] > best:
            best, best_ep, wait = mv[MONITOR], ep, 0
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
        else:
            wait += 1
            if wait >= PATIENCE:
                print(f"  [fold {fold}] early stop @ ep {ep} (best {best:.3f} @ ep {best_ep})"); break
    model.load_state_dict(best_state)
    yv, pv = predict(model, dl_va)
    return hist, full_metrics(yv, pv), best_ep, pv


# ===== CELL 5: 5-Fold Stratified-Group CV =====
sgkf = StratifiedGroupKFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
oof_pred = -np.ones(N, dtype=np.int64); all_history, fold_metrics, best_epochs = [], [], []
for fold, (idx_tr, idx_va) in enumerate(sgkf.split(X.reshape(N, -1), y, groups)):
    assert set(groups[idx_tr]).isdisjoint(set(groups[idx_va]))
    print(f"\n=== FOLD {fold} | train {len(idx_tr)} / val {len(idx_va)} ===")
    hist, mets, best_ep, pv = train_one_fold(fold, idx_tr, idx_va)
    oof_pred[idx_va] = pv; all_history += hist
    fold_metrics.append({"fold": fold, **{k: round(mets[k], 4) for k in METRIC_COLS}, "best_epoch": best_ep})
    best_epochs.append(best_ep)
    print(f"--- Fold {fold}: macroF1 {mets['macroF1']:.3f} | recall_jatuh {mets['recall_jatuh']:.3f}")
assert (oof_pred >= 0).all()


# ===== CELL 6: Hasil CV + simpan + grafik + model final =====
fold_df = pd.DataFrame(fold_metrics)
print("=== Metrik per FOLD (ST-GCN) ===\n" + fold_df.to_string(index=False))
summary = {k: {"mean": round(float(fold_df[k].mean()), 4), "std": round(float(fold_df[k].std(ddof=0)), 4)}
           for k in METRIC_COLS}
print("\n=== Ringkasan lintas fold (mean ± std) ===")
for k in METRIC_COLS: print(f"  {k:14s}: {summary[k]['mean']:.3f} ± {summary[k]['std']:.3f}")

cm = confusion_matrix(y, oof_pred, labels=[0, 1, 2])
print("\nConfusion matrix OOF [normal, oleng, jatuh]:\n", cm)
print(classification_report(y, oof_pred, labels=[0,1,2], target_names=CLASS_NAMES, zero_division=0, digits=3))

fold_df.to_csv(f"{MODEL_DIR}/fall_{TAG}_cv_folds.csv", index=False)
pd.DataFrame(all_history).to_csv(f"{MODEL_DIR}/fall_{TAG}_cv_history.csv", index=False)
with open(f"{MODEL_DIR}/fall_{TAG}_cv_summary.json", "w") as f:
    json.dump({"arch": "STGCN", "n_folds": N_FOLDS, "summary_mean_std": summary,
               "oof_confusion_matrix": cm.tolist()}, f, indent=2)
hist_df = pd.DataFrame(all_history)
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].bar(fold_df.fold-0.15, fold_df.macroF1, 0.3, label="macroF1")
ax[0].bar(fold_df.fold+0.15, fold_df.recall_jatuh, 0.3, label="recall_jatuh")
ax[0].set_ylim(0,1); ax[0].set_title("ST-GCN metrik per fold"); ax[0].legend()
for fk in sorted(hist_df.fold.unique()):
    s = hist_df[hist_df.fold==fk]; ax[1].plot(s.epoch, s.macroF1, marker='.', label=f"fold {fk}")
ax[1].set_ylim(0,1); ax[1].set_title("Learning curve"); ax[1].legend()
plt.tight_layout(); plt.savefig(f"{MODEL_DIR}/fall_{TAG}_cv_curves.png", dpi=120); plt.show()

# model final di semua data
final_epochs = max(3, int(round(np.mean(best_epochs))))
print(f"\nModel FINAL ST-GCN, {final_epochs} epoch, semua data.")
cnt = np.bincount(y, minlength=N_CLASSES).astype(np.float32); cw = cnt.sum()/(N_CLASSES*np.clip(cnt,1,None))
final = make_model(); crit = nn.CrossEntropyLoss(weight=torch.tensor(cw, device=DEVICE))
opt = torch.optim.Adam(final.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
dl_all = make_loader(np.arange(N), True, drop_last=True)
for ep in range(1, final_epochs+1):
    final.train(); run=0; seen=0
    for xb, yb in tqdm(dl_all, desc=f"FINAL {ep:02d}/{final_epochs}", leave=False):
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        opt.zero_grad(); loss = crit(final(xb), yb); loss.backward()
        torch.nn.utils.clip_grad_norm_(final.parameters(), GRAD_CLIP); opt.step()
        run += loss.item()*len(yb); seen += len(yb)
    print(f"  FINAL ep {ep:02d} | loss {run/max(seen,1):.4f}")
torch.save(final.state_dict(), f"{MODEL_DIR}/fall_head_{TAG}.pt")
with open(f"{MODEL_DIR}/fall_head_{TAG}.json", "w") as f:
    json.dump({"arch": "STGCN", "in_channels": USE_CHANNELS, "hidden": HIDDEN,
               "n_classes": N_CLASSES, "class_names": CLASS_NAMES, "fall_joints": FALL_JOINTS,
               "edges": FALL_EDGES, "use_channels": USE_CHANNELS, "window": 45, "stride": 15,
               "fps": 15, "final_epochs": final_epochs, "cv_macroF1": summary["macroF1"],
               "cv_recall_jatuh": summary["recall_jatuh"]}, f, indent=2)
print(f"Tersimpan: fall_head_{TAG}.pt + fall_head_{TAG}.json + fall_{TAG}_cv_*")


# ===== CELL 7: Efisiensi + PERBANDINGAN dgn BiLSTM =====
n_params = sum(p.numel() for p in final.parameters())
final.eval(); xb = torch.tensor(X[:min(BATCH, N)]).to(DEVICE)
with torch.no_grad():
    for _ in range(3): final(xb)
    if DEVICE == "cuda": torch.cuda.synchronize()
    t0 = time.time()
    for _ in range(20): final(xb)
    if DEVICE == "cuda": torch.cuda.synchronize()
    ms = (time.time()-t0)/20/xb.shape[0]*1000
print(f"ST-GCN: {n_params:,} params | ~{ms:.3f} ms/jendela ({DEVICE})")

print("\n================ PERBANDINGAN BiLSTM vs ST-GCN ================")
row = f"{'metrik':16s}| {'BiLSTM':>18s} | {'ST-GCN':>18s}"
print(row); print("-"*len(row))
bilstm_path = f"{MODEL_DIR}/fall_cv_summary.json"
if os.path.exists(bilstm_path):
    b = json.load(open(bilstm_path))["summary_mean_std"]
    for k in METRIC_COLS:
        bs = f"{b[k]['mean']:.3f} ± {b[k]['std']:.3f}"
        ss = f"{summary[k]['mean']:.3f} ± {summary[k]['std']:.3f}"
        print(f"{k:16s}| {bs:>18s} | {ss:>18s}")
    print("\n(Ambil tabel ini untuk bab 'Kematangan Arsitektur' di proposal.)")
else:
    print("fall_cv_summary.json (BiLSTM) belum ada -> jalankan train_fall.py dulu utk pembanding.")
    for k in METRIC_COLS:
        print(f"{k:16s}| {'(-)':>18s} | {summary[k]['mean']:.3f} ± {summary[k]['std']:.3f}")
```
